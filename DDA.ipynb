```python
import matplotlib.pyplot as plt
```

```python
def dda_line(x1, y1, x2, y2):
  print("\n ---DDA Line Drawing Algorithm--- ")
  print("Starting point: ",x1,y1)
  print("Ending point: ",x2,y2)
  dx = x2 - x1
  dy = y2 - y1
  # Conditions
  if abs(dx) >= abs(dy):
        print("Case 1: dx=1")
  else:
        print("Case 2: dy=1")

  # Number of steps = greater of |dx| or |dy|
  steps = int(max(abs(dx), abs(dy)))

  # Avoid division by zero when start == end
  if steps == 0:
      return [(round(x1), round(y1))]

  # Increment in x and y for each step
  x_inc = dx / steps
  y_inc = dy / steps

  x, y = x1, y1
  points = []

  for _ in range(steps + 1):
      points.append((round(x), round(y)))
      x += x_inc
      y += y_inc
  return points
```

```python
def plot_line(points, x1, y1, x2, y2):
    xs = [p[0] for p in points]
    ys = [p[1] for p in points]

    plt.figure(figsize=(7, 6))
    plt.plot(xs, ys, color="blue", linewidth=1, label="Line")
    plt.scatter(xs, ys, color="red", s=25, zorder=5, label="Computed points")
    # Plot the ideal mathematical line for comparison
    plt.plot([x1, x2], [y1, y2], color="magenta", linestyle="--", linewidth=1, label="Ideal line")
    # Highlight the original start/end pixels
    plt.scatter([x1, x2], [y1, y2], color="green", s=80,
                marker="o", zorder=6, label="Start / End")

    for x,y in points: # Text annotation (Coordinates)
     plt.text(x + 0.08, y + 0.08, f"({x},{y})", fontsize=9)
    plt.title("DDA Line Drawing Algorithm")
    plt.xlabel("X")
    plt.ylabel("Y")
    plt.grid(True)
    plt.legend()
    plt.axis('equal')
    plt.show()


if __name__ == "__main__":

  #Enter Coordinates
  x1 = int(input("Enter x1: "))
  y1 = int(input("Enter y1: "))
  x2 = int(input("Enter x2: "))
  y2 = int(input("Enter y2: "))

  result = dda_line(x1, y1, x2, y2)
  print("Points on the line:")
  for p in result:
        print(p)

  plot_line(result, x1, y1, x2, y2)
```
